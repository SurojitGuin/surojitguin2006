```python
# ==========================================
# CELL 1: Environment Setup
# ==========================================
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Set random seed for reproducibility
np.random.seed(42)

# Configure plot appearance
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True

print("Libraries successfully loaded and environment configured.")
```

```python
# ==========================================
# CELL 2: PDF and CDF Estimation (N = 10^5)
# ==========================================
N = 100_000

# 1. Generate Uniform Random Variable U ~ Uniform(a=-2, b=2)
a, b = -2.0, 2.0
uniform_samples = np.random.uniform(a, b, N)

# 2. Generate Gaussian Random Variable X ~ Normal(mu=0, sigma=1)
mu, sigma = 0.0, 1.0
gaussian_samples = np.random.normal(mu, sigma, N)

# Compute moments
print(f"--- Uniform Distribution [a={a}, b={b}] ---")
print(f"Theoretical Mean: {(a+b)/2:.4f} | Sample Mean: {np.mean(uniform_samples):.4f}")
print(f"Theoretical Var : {((b-a)**2)/12:.4f} | Sample Var : {np.var(uniform_samples, ddof=1):.4f}\n")

print(f"--- Gaussian Distribution [mu={mu}, sigma={sigma}] ---")
print(f"Theoretical Mean: {mu:.4f} | Sample Mean: {np.mean(gaussian_samples):.4f}")
print(f"Theoretical Var : {sigma**2:.4f} | Sample Var : {np.var(gaussian_samples, ddof=1):.4f}")

# Plotting PDF Comparisons
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Uniform PDF Plot
count, bins, _ = axes[0].hist(uniform_samples, bins=60, density=True, alpha=0.6, color='tab:blue', label='Empirical Histogram')
u_theory = np.where((bins >= a) & (bins <= b), 1.0 / (b - a), 0.0)
axes[0].plot(bins, u_theory, 'r--', lw=2.5, label='Theoretical PDF')
axes[0].set_title("Uniform Distribution: Normalized Histogram vs. PDF")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Probability Density")
axes[0].legend()

# Gaussian PDF Plot
count, bins, _ = axes[1].hist(gaussian_samples, bins=60, density=True, alpha=0.6, color='tab:green', label='Empirical Histogram')
g_theory = (1.0 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((bins - mu) / sigma)**2)
axes[1].plot(bins, g_theory, 'r--', lw=2.5, label='Theoretical PDF')
axes[1].set_title("Gaussian Distribution: Normalized Histogram vs. PDF")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Probability Density")
axes[1].legend()

plt.tight_layout()
plt.show()

# Plotting Empirical CDFs
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

# Uniform CDF
u_sorted = np.sort(uniform_samples)
cdf_u_emp = np.arange(1, N + 1) / N
cdf_u_theory = np.clip((u_sorted - a) / (b - a), 0, 1)

axes[0].plot(u_sorted, cdf_u_emp, label='Empirical CDF', color='tab:blue')
axes[0].plot(u_sorted, cdf_u_theory, 'r--', lw=2, label='Theoretical CDF')
axes[0].set_title("Uniform Distribution CDF")
axes[0].set_xlabel("x")
axes[0].set_ylabel("F(x)")
axes[0].legend()

# Gaussian CDF
g_sorted = np.sort(gaussian_samples)
cdf_g_emp = np.arange(1, N + 1) / N
cdf_g_theory = stats.norm.cdf(g_sorted, mu, sigma)

axes[1].plot(g_sorted, cdf_g_emp, label='Empirical CDF', color='tab:green')
axes[1].plot(g_sorted, cdf_g_theory, 'r--', lw=2, label='Theoretical CDF')
axes[1].set_title("Gaussian Distribution CDF")
axes[1].set_xlabel("x")
axes[1].set_ylabel("F(x)")
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 3: Convergence Analysis across Sample Sizes
# ==========================================
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

sample_sizes = [100, 1000, 10000, 100000, 1000000]
mu_true = 0.0
sigma_true = 1.0
var_true = 1.0

# Generate datasets without indented for-loops
samples = [np.random.normal(mu_true, sigma_true, n) for n in sample_sizes]
means = [float(np.mean(s)) for s in samples]
variances = [float(np.var(s, ddof=1)) for s in samples]
abs_errors_mean = [abs(m - mu_true) for m in means]
abs_errors_var = [abs(v - var_true) for v in variances]

print(f"{'Sample Size (N)':<18} | {'Est. Mean':<12} | {'Est. Variance':<14} | {'Mean Error':<12}")
print("-" * 62)
_ = [print(f"{n:<18} | {m:<12.6f} | {v:<14.6f} | {e:<12.6f}") for n, m, v, e in zip(sample_sizes, means, variances, abs_errors_mean)]

plt.figure(figsize=(8, 4.5))
plt.loglog(sample_sizes, abs_errors_mean, marker='o', color='crimson', lw=2, label='|Mean Error|')
plt.title("Convergence of Sample Mean Error vs. Sample Size (N)")
plt.xlabel("Sample Size (N)")
plt.ylabel("Absolute Error")
plt.grid(True, which="both", ls="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 4: AWGN Injection & SNR Measurement
# ==========================================
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 1. Generate clean sinusoid (200 Hz, 10 ms duration)
fs = 10000
t = np.linspace(0, 0.01, int(fs * 0.01), endpoint=False)
amplitude = 2.0
f_sig = 200.0
clean_signal = amplitude * np.sin(2.0 * np.pi * f_sig * t)

# 2. AWGN calculation parameters
target_snr_db = 10.0
p_signal = float(np.mean(clean_signal ** 2))
snr_linear = 10.0 ** (target_snr_db / 10.0)
p_noise_target = p_signal / snr_linear

# 3. Generate noise and noisy signal
noise = np.random.normal(0.0, np.sqrt(p_noise_target), len(clean_signal))
noisy_signal = clean_signal + noise

# 4. Measure resulting empirical SNR
p_noise_measured = float(np.mean(noise ** 2))
measured_snr_db = float(10.0 * np.log10(p_signal / p_noise_measured))

# 5. Display comparison
print(f"Target SNR   : {target_snr_db:.2f} dB")
print(f"Measured SNR : {measured_snr_db:.2f} dB")
print(f"Discrepancy  : {abs(target_snr_db - measured_snr_db):.4f} dB")

# 6. Plot waveforms
plt.figure(figsize=(10, 4.5))
plt.plot(t * 1e3, clean_signal, 'b-', lw=2.5, label='Clean Signal (200 Hz)')
plt.plot(t * 1e3, noisy_signal, 'r--', alpha=0.75, label=f'Noisy Signal (SNR = {measured_snr_db:.2f} dB)')
plt.title("Waveform Comparison: Clean vs. AWGN-Corrupted Sinusoid")
plt.xlabel("Time (ms)")
plt.ylabel("Amplitude (V)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 5: Noise Autocorrelation Function
# ==========================================
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 1. Generate a zero-mean Gaussian noise sequence
num_samples = 20000
noise_sequence = np.random.normal(0.0, 1.0, num_samples)

# 2. Compute full discrete autocorrelation and normalize
autocorr = np.correlate(noise_sequence, noise_sequence, mode='full')
autocorr = autocorr / num_samples

# 3. Extract central lag window [-50, +50]
lags = np.arange(-num_samples + 1, num_samples)
center_idx = len(autocorr) // 2
span = 50
lags_subset = lags[center_idx - span : center_idx + span + 1]
autocorr_subset = autocorr[center_idx - span : center_idx + span + 1]

# 4. Plot empirical autocorrelation
plt.figure(figsize=(10, 4.5))
plt.stem(lags_subset, autocorr_subset, basefmt=" ")
plt.title("Empirical Autocorrelation of AWGN Sequence")
plt.xlabel("Lag (k)")
plt.ylabel("R_ww[k]")
plt.axhline(0, color='black', lw=1)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
